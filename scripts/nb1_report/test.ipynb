# Meta: league, season, round_number, date_range, logo_path

```python
# Import scraper, fetch standings
tour_id = 187 # Fizz Liga
season_id = 76944
def fetch_standings(tour_id, season_id):
    import os 
    from pathlib import Path 
    import sys 

    wd = Path(os.getcwd())
    root = wd.parents[1]
    sys.path.append(f"{root}/modules")

    from SofaScore_module import scrape_sofascore



    url_stand = f"https://www.sofascore.com/api/v1/unique-tournament/{tour_id}/season/{season_id}/standings/total"
    stand = scrape_sofascore(url_stand)['standings'][0]

    return stand 

stand = fetch_standings(tour_id, season_id)
```

```python
# Fetch matches

def fetch_matches(tour_id, season_id, round):
    url_matches = f"https://www.sofascore.com/api/v1/unique-tournament/{tour_id}/season/{season_id}/events/round/{round}"

    matches = scrape_sofascore(url_matches)

    return matches

matches = fetch_matches(tour_id, season_id, round=14)
```

```python
def extract_meta(tour_id, season_id, round):
    stand = fetch_standings(tour_id, season_id)
    matches = fetch_matches(tour_id, season_id, round) 
    
    # league
    league = stand['tournament']['uniqueTournament']['name']

    # season 
    season = stand['name'].split(' ')[-1]

    # round_number 
    matches_max = 0
    for row in stand['rows']:
        if row['matches'] > matches_max:
            matches_max = row['matches']

    # date_range 
    min_date = 1e10000
    max_date = 0
    for event in matches['events']:
        if event['startTimestamp'] < min_date:
            min_date = event['startTimestamp']
        if event['startTimestamp'] > max_date:
            max_date = event['startTimestamp']
            
    import time
    formatted_time_min = time.strftime('%Y-%m-%d %H:%M:%S', time.gmtime(min_date))
    formatted_time_max = time.strftime('%Y-%m-%d %H:%M:%S', time.gmtime(max_date))

    # logo_path
    logo_path = f"https://img.sofascore.com/api/v1/unique-tournament/{tour_id}/image"

    return {
        "league_name": league,
        "season": season,
        "round_number": matches_max,
        "date_range": {
            "start": formatted_time_min,
            "end": formatted_time_max
        },
        "logo_path": logo_path
    }

meta = extract_meta(tour_id, season_id, round=14)
print(meta)
```
